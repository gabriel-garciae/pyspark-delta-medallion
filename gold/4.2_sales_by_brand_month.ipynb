Libraries imports

```python
import pyspark.sql.functions as F
```

Read and transformations

```python
df_sales = spark.table("abinbev_case_silver.fact_sales")

df_sales = (
    df_sales
    .withColumnRenamed("year_partition", "year")
    .withColumnRenamed("month_partition", "month")
)

df_sales_by_brand_month = (
    df_sales
    .groupBy("year", "month", "brand_nm")
    .agg(F.sum("volume").alias("total_volume"))
    .orderBy("year", "month", "brand_nm")
)
```

Save as gold delta table

```python
df_sales_by_brand_month.write.mode("overwrite").format("delta").saveAsTable(
    "abinbev_case_gold.v_sales_by_brand_month"
)

display(df_sales_by_brand_month)
```
