Libraries imports

```python
import pyspark.sql.functions as F
```

Read and light transformations

```python
raw_dir = "dbfs:/FileStore/tables"
file_channel = "abi_bus_case1_beverage_channel_group_20210726.csv"

df_channel = (
    spark.read.format("csv")
        .option("header", "true")
        .option("encoding", "UTF-8")
        .option("inferSchema", "true")
        .option("sep", ",")
        .load(f"{raw_dir}/{file_channel}")
)

display(df_channel)
print(f"Total rows in channel group dataset: {df_channel.count()}")
```

Save as bronze delta table

```python
df_channel.write.mode("overwrite").format("delta").saveAsTable(
    "abinbev_case_bronze.beverage_channel_group")
```
