Libraries imports

```python
import pyspark.sql.functions as F
```

Read and transformations

```python
df_sales = spark.table("abinbev_case_bronze.beverage_sales")

df_dim_brand = (
    df_sales
    .select(F.trim(F.upper(F.col("brand_nm"))).alias("brand_nm"))
    .dropDuplicates()
    .withColumn("brand_id", F.abs(F.crc32(F.col("brand_nm"))))
    .select("brand_id", "brand_nm")
)
```

Save as silver delta table

```python
df_dim_brand.write.mode("overwrite").format("delta").saveAsTable(
    "abinbev_case_silver.dim_brand")

display(df_dim_brand)
```
