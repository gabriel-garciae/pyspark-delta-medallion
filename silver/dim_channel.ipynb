Libraries imports

```python
import pyspark.sql.functions as F
```

Read and transformations

```python
df_channel = spark.table("abinbev_case_bronze.beverage_channel_group")

df_dim_channel = (
    df_channel
    .select(
        F.trim(F.upper(F.col("TRADE_CHNL_DESC"))).alias("trade_chnl_desc"),
        F.trim(F.upper(F.col("TRADE_GROUP_DESC"))).alias("trade_group_desc"),
        F.trim(F.upper(F.col("TRADE_TYPE_DESC"))).alias("trade_type_desc")
    )
    .dropDuplicates()
)
```

Save as silver delta table

```python
df_dim_channel.write.mode("overwrite").format("delta").saveAsTable(
    "abinbev_case_silver.dim_channel")

display(df_dim_channel)
```
